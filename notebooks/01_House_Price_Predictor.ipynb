```python
from src.prediction import AhmedabadPricePredictor

predictor = AhmedabadPricePredictor()

sample_property = {
    "locality": "Shela",
    "bhk": 3,
    "total_sqft": 1500,
    "type_area": "Carpet Area",
    "status": "Ready to Move",
    "furnishing": "Semi-Furnished",
    "transaction": "Resale"
}

result = predictor.predict(sample_property)

print("===== AI HOUSE PRICE PREDICTION =====")
print("Location:", sample_property["locality"])
print("BHK:", sample_property["bhk"])
print("Area:", sample_property["total_sqft"], "sq.ft")
print("Predicted Price:", result["formatted_rupees"])
print("Price Range:", result["lower_bound_formatted"],
      "to", result["upper_bound_formatted"])
print("Rate per Sq.ft:", result["rate_per_sqft_formatted"])
print("Model:", result["model_name"])
```
